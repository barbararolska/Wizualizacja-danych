## Zadanie 0: Generowanie danych do animacji
---

Aby uczyć się animacji, potrzebujemy danych posiadających wymiar czasu. Zamiast szukać gotowego pliku, napiszemy krótki kod, który zasymuluje ruch 5 obiektów (np. samochodów dostawczych) na terytorium Polski przez 60 "kroków" (np. minut).

Każdy obiekt będzie miał:
* `id` - identyfikator pojazdu.
* `step` - krok w czasie (nasza "klatka" animacji).
* `lon` (Długość geograficzna - X) oraz `lat` (Szerokość geograficzna - Y) - współrzędne w granicach Polski, zmieniające się w czasie na zasadzie błądzenia losowego (Random Walk).
* `value` - symulowana wartość (np. prędkość pojazdu), która będzie się płynnie zmieniać.

**Instrukcja:**
Wystarczy, że uruchomisz poniższą komórkę. Zrozumienie kodu generującego nie jest głównym celem zajęć, ale rzuć okiem na strukturę wygenerowanej tabeli (obiekt `df_sim`), ponieważ na niej opierać się będą wszystkie kolejne zadania.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# ZADANIE 0: Generowanie symulowanych danych
# ==========================================

np.random.seed(42) # Ustawiamy ziarno dla powtarzalności wyników

n_agents = 5   # Liczba śledzonych obiektów (np. samochodów)
n_steps = 60   # Liczba klatek w naszej animacji

data = []

# Symulujemy każdego agenta po kolei
for agent_id in range(n_agents):
    # Losujemy punkt startowy gdzieś w centrum Polski (Długość ~19, Szerokość ~52)
    start_lon = np.random.uniform(16.0, 22.0)
    start_lat = np.random.uniform(50.0, 53.0)
    start_val = np.random.uniform(30, 80)

    lon, lat, val = start_lon, start_lat, start_val

    for step in range(n_steps):
        data.append([agent_id, step, lon, lat, val])

        # Pojazd porusza się losowo o niewielki wektor w każdym kroku
        lon += np.random.normal(0, 0.05)
        lat += np.random.normal(0, 0.05)
        # Prędkość lekko rośnie lub maleje
        val += np.random.normal(0, 3.0)

# Zapisujemy do DataFrame
df_sim = pd.DataFrame(data, columns=['id', 'step', 'lon', 'lat', 'value'])

print("Struktura wygenerowanych danych:")
display(df_sim.head())
```

## Zadanie 1: Podstawowa animacja w Matplotlib
---
Czas na pierwszą animację! Zasada działania modułu `FuncAnimation` z `matplotlib.animation` opiera się na **funkcji aktualizującej** (tzw. `update`).

**Jak to działa?**
1. Najpierw przygotowujesz puste płótno (`fig, ax`) i rysujesz "startowy" stan wykresu (np. puste kropki, albo kropki z pierwszego kroku). Wykres ten przypisujesz do zmiennej (np. `scatter_plot`).
2. Definiujesz w Pythonie własną funkcję `def update(frame):`. Słowo `frame` to numer klatki (od 0 do 59 w naszym przypadku).
3. Wewnątrz funkcji `update` mówisz programowi: *"Dla klatki numer X, odfiltruj z tabeli wiersze, w których `step == X`, a następnie zmień pozycje punktów na wykresie na te odfiltrowane"*.
4. Uruchamiasz `FuncAnimation()`, która sama w pętli wywołuje Twoją funkcję `update()` tyle razy, ile jest klatek.
5. W Google Colab animację wyświetlamy, korzystając z biblioteki `IPython.display`, która zamienia ją na interaktywne wideo w HTML.

**Kluczowe funkcje i argumenty:**
* `ax.scatter(X, Y)` – rysuje kropki. Przypisz ją do zmiennej, np. `punkty = ax.scatter(...)`.
* `punkty.set_offsets(tablica_XY)` – najszybsza metoda animacji! Zamiast usuwać i rysować wykres od nowa (co muli komputer), ta funkcja bierze narysowane już kropki i po prostu **przesuwa** je na nowe koordynaty. Tablica XY musi być strukturą o dwóch kolumnach (np. wyciągniętą z `df[['lon', 'lat']].values`).
* `FuncAnimation(fig, update, frames=60, interval=100)`:
   * `fig` - płótno, na którym rysujemy.
   * `update` - nazwa naszej funkcji aktualizującej (zauważ, bez nawiasów po nazwie!).
   * `frames` - ile łącznie klatek ma wygenerować.
   * `interval` - czas między klatkami w milisekundach (100 ms = 10 klatek na sekundę).
* `HTML(anim.to_jshtml())` – renderuje kontroler wideo wewnątrz Colaba.

**Instrukcja krok po kroku:**
1. Zaimportuj `from matplotlib.animation import FuncAnimation` oraz `from IPython.display import HTML`.
2. Zainicjuj wykres: `fig, ax = plt.subplots(figsize=(8, 6))`.
3. Ustaw stałe limity osi za pomocą `ax.set_xlim(15, 23)` i `ax.set_ylim(49, 55)`, by wykres nie "skakał" przy każdym ruchu!
4. Wyciągnij z `df_sim` dane tylko dla `step == 0` i narysuj punkt startowy: `punkty = ax.scatter(dane['lon'], dane['lat'], s=100, c=dane['id'], cmap='tab10')`. Zauważ kolorowanie względem 'id', by odróżnić samochody.
5. Stwórz funkcję `def update(frame):`. Wewnątrz:
   * Dodaj `ax.set_title(...)` ze zmiennym napisem podającym aktualny `frame` (użyj f-stringa).
   * Odfiltruj dane: zostaw tylko wiersze, gdzie `df_sim['step'] == frame`.
   * Stwórz dwukolumnową tablicę ze współrzędnymi: `nowe_pozycje = aktualne_dane[['lon', 'lat']].values`.
   * Przesuń istniejące kropki: `punkty.set_offsets(nowe_pozycje)`.
   * Pamiętaj! Funkcja powinna na końcu zwrócić obiekt punktów: `return punkty,` (przecinek na końcu jest wymogiem matplotlib dla optymalizacji, chociaż Colab wybacza jego brak).
6. Zamknij wykres wywołując statyczne `plt.close()` – aby Colab nie narysował pustego wykresu pod filmem.
7. Wygeneruj animację `anim = FuncAnimation(...)` z 60 klatkami i interwałem 100.
8. Wyświetl: `HTML(anim.to_jshtml())`. Kliknij Play!

```python
#@title Miejsce na Twój kod
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlim(15, 23)
ax.set_ylim(49, 55)
dane = df_sim[df_sim['step'] == 0]
punkty = ax.scatter(dane['lon'], dane['lat'], s=100, c=dane['id'], cmap='tab10')
def update(frame):
  ax.set_title(f'Klatka: {frame}')
  aktualne_dane = df_sim[df_sim['step'] == frame]
  nowe_pozycje = aktualne_dane[['lon', 'lat']].values
  punkty.set_offsets(nowe_pozycje)
  return punkty,
plt.close()
anim = FuncAnimation(fig, update, frames=60, interval=100)
HTML(anim.to_jshtml())
```

## Zadanie 2: Animacja na wykresie Seaborn (Aktualizacja przez "Czyszczenie")
---

O ile w `matplotlib` mogliśmy płynnie przesuwać kropki funkcją `.set_offsets()`, o tyle statystyczna biblioteka `seaborn` (np. wykresy słupkowe czy gęstości) rysuje bardzo skomplikowane i wielowarstwowe kształty, których nie da się łatwo "przesunąć".

Dlatego dla Seaborn stosujemy technikę **"Clear & Redraw"** (Wyczyść i Narysuj na nowo).

**Co musisz wiedzieć o tym podejściu:**
* Wewnątrz funkcji `update(frame)` jako pierwszą komendę wpisujemy `ax.clear()`. Zmazuje ona wszystko, co było na wykresie w poprzedniej klatce (nawet limity osi i tytuły!).
* Następnie rysujemy nowy wykres wywołując normalnie `sns.histplot(..., ax=ax)` dla odfiltrowanych danych z danego kroku.
* Ponieważ `ax.clear()` usuwa WSZYSTKO, musimy w każdej klatce od nowa deklarować granice `ax.set_xlim(...)` oraz tytuły `ax.set_title(...)`. Jest to wolniejsza metoda komputerowo, ale niezbędna dla złożonych wykresów z Seaborn.

**Instrukcja krok po kroku:**
Zrobimy animację rozkładu (histogramu) prędkości (`value`) dla wszystkich 5 aut na każdym kroku czasu.
0. Zaimportuj paczkę seaborn `import seaborn as sns`
1. Zainicjuj `fig, ax = plt.subplots(...)`.
2. Zbuduj `def update(frame):`
3. Wewnątrz wywołaj `ax.clear()`. To absolutnie kluczowe!
4. Wyfiltruj `aktualne_dane = df_sim[df_sim['step'] == frame]`.
5. Narysuj `sns.histplot` (bądź `kdeplot`), przekazując `data=aktualne_dane`, `x='value'`, `ax=ax`, `color='skyblue'`, `kde=True`.
6. Od nowa nadaj limity osi X: `ax.set_xlim(10, 100)` (żeby oś X nie skakała) oraz oś Y `ax.set_ylim(0, 5)`.
7. Nadaj też od nowa tytuł wykresu.
8. Wywołaj `FuncAnimation(...)` na 60 klatkach z interwałem 150ms.
9. Pamiętaj o `plt.close()` i wywołaj `HTML(anim.to_jshtml())`. Zobacz jak ładnie "oddycha" rozkład danych w czasie!

```python
#@title Miejsce na Twój kod
import seaborn as sns
fig, ax = plt.subplots(figsize=(8, 6))
def update(frame):
  ax.clear()
  aktualne_dane = df_sim[df_sim['step'] == frame]
  sns.histplot(data=aktualne_dane, x='value', ax=ax, color='skyblue', kde=True)
  ax.set_xlim(10, 100)
  ax.set_ylim(0, 5)
  ax.set_title(f'Rozkład parametru - Klatka: {frame}')

plt.close()
anim = FuncAnimation(fig, update, frames=60, interval=150)
HTML(anim.to_jshtml())
```

## Zadanie 3: Geopandas – Ruchoma mapa

Mamy już podstawy animacji punktów i znamy podstawy Geopandas z poprzednich zajęć. Połączmy to!

**Instrukcja krok po kroku:**
1. Importuj `geopandas as gpd`. Wczytaj kontur Polski z GitHuba:

```python
#@title Miejsce na Twój kod
import geopandas as gpd
url = "https://raw.githubusercontent.com/codeforamerica/click_that_hood/master/public/data/poland.geojson"
polska = gpd.read_file(url)
fig, ax = plt.subplots(figsize=(10, 10))
polska.plot(ax=ax, color='whitesmoke', edgecolor='grey')
ax.set_axis_off()
pojazdy = ax.scatter(df_sim[df_sim['step'] == 0]['lon'], df_sim[df_sim['step'] == 0]['lat'], zorder=5)
def update(frame):
  ax.set_title(f'Klatka: {frame}')
  aktualne_dane = df_sim[df_sim['step'] == frame]
  nowe_pozycje = aktualne_dane[['lon', 'lat']].values
  pojazdy.set_offsets(nowe_pozycje)
  return pojazdy,

plt.close()
anim = FuncAnimation(fig, update, frames=60, interval=100)
HTML(anim.to_jshtml())
```

## Zadanie A (Dla chętnych): Zostawianie "Ogonków" (Worm trails / Trajectories)
---

Sam skaczący punkt często nie pozwala prześledzić całej trasy (trajektorii) pojazdu, jeśli widz mrugnie. Profesjonalne animacje na mapach rozwiązują to, rysując w tle tzw. ogonki (historie trasy do danego momentu).

W tym zadaniu spróbujemy połączyć wyczyszczenie całego płótna w Seaborn z rysowaniem trajektorii.

**Pomysł na algorytm:**
W funkcji `update(frame)`, zamiast pobierać tylko wiersze, gdzie `step == frame`, pobierzemy wszystkie wiersze, gdzie `step <= frame` (cała historia do tej chwili włącznie!).
Następnie wyrysujemy je na zwykłym wykresie korzystając np. z `sns.lineplot()`.
Ponieważ `lineplot` sam "łączy kropki" wedle jakiejś kategorii (u nas kolumna `id`), w każdym kolejnym kroku trasa linii będzie rosła! Na sam koniec, nad linią dorysujemy duży marker na ostatniej, aktualnej pozycji.

**Instrukcja krok po kroku:**
1. Stwórz `fig, ax` z zamrożonymi osiami dla Polski.
2. Zdefiniuj `update(frame)`:
   * Wywołaj `ax.clear()`.
   * Odfiltruj `historia = df_sim[df_sim['step'] <= frame]`.
   * Z tej `historii` wyciągnij tylko `aktualne = historia[historia['step'] == frame]` na potrzeby kropki końcowej.
   * Użyj `sns.lineplot(data=historia, x='lon', y='lat', hue='id', ax=ax, linewidth=3, legend=False)`. Funkcja ta sama połączy odpowiednie id w linie!
   * Skopiuj punkt "na wierzch": `ax.scatter(aktualne['lon'], aktualne['lat'], c='black', s=100, zorder=5)`.
   * Zamroź ponownie limity (`xlim`, `ylim`).
3. Zbuduj i wyświetl animację. Patrz, jak "robaczki" malują swoje trasy!

```python
#@title Miejsce na Twój kod
fig, ax = plt.subplots(figsize=(10, 10))
polska.plot(ax=ax, color='whitesmoke', edgecolor='grey')
ax.set_axis_off()
ax.set_xlim(15, 23)
ax.set_ylim(49, 55)
def update(frame):
  ax.set_title(f'Kreślenie trajektorii - krok: {frame}')
  ax.clear()
  historia = df_sim[df_sim['step'] <= frame]
  aktualne = historia[historia['step'] == frame]
  sns.lineplot(data=historia, x='lon', y='lat', hue='id', ax=ax, linewidth=3, legend=False)
  ax.scatter(aktualne['lon'], aktualne['lat'], c='black', s=100, zorder=5)
  ax.set_xlim(15, 23)
  ax.set_ylim(49, 55)
plt.close()
anim = FuncAnimation(fig, update, frames=60, interval=100)
HTML(anim.to_jshtml())
```
